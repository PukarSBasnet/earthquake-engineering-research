# 1-D Site Response of Kathmandu Valley Boreholes

Shear-wave velocity (Vs) profiles were estimated from SPT-N values using the **Ohta & Goto (1978)** and **Imai & Tonouchi (1982)** correlations. For each borehole, linear amplification functions (rock-outcrop, borehole and incident motions) are computed with [PySeismoSoil](https://github.com/jsh9/PySeismoSoil).

Profile columns: thickness (m), Vs (m/s), damping (%), density (g/cm³), material number. All profiles are also collected in `data/vs_profiles.csv`.
